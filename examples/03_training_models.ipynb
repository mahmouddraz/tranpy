# Training Models

## Setup

```python
from tranpy.datasets import load_newengland

X_train, X_test, y_train, y_test = load_newengland(
    test_size=0.2,
    random_state=42
)
```

## Decision Tree

```python
from tranpy.models import DecisionTreeClassifier

model = DecisionTreeClassifier(max_depth=10)
model.fit(X_train, y_train)

results = model.evaluate(X_test, y_test, verbose=False)
print(f"Accuracy: {results['accuracy']:.3f}")
```

## Random Forest

```python
from tranpy.models import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100, max_depth=10)
model.fit(X_train, y_train)

results = model.evaluate(X_test, y_test, verbose=False)
print(f"Accuracy: {results['accuracy']:.3f}")
```

## SVM

```python
from tranpy.models import SVMClassifier

model = SVMClassifier(kernel='rbf', C=1.0)
model.fit(X_train[:1000], y_train[:1000])

results = model.evaluate(X_test[:200], y_test[:200], verbose=False)
print(f"Accuracy: {results['accuracy']:.3f}")
```

## Neural Network (MLP)

```python
from tranpy.models import MLPClassifier

model = MLPClassifier(
    hidden_layer_sizes=(100, 50),
    max_iter=200
)
model.fit(X_train, y_train)

results = model.evaluate(X_test, y_test, verbose=False)
print(f"Accuracy: {results['accuracy']:.3f}")
```

## Logistic Regression

```python
from tranpy.models import LogisticRegression

model = LogisticRegression(max_iter=500)
model.fit(X_train, y_train)

results = model.evaluate(X_test, y_test, verbose=False)
print(f"Accuracy: {results['accuracy']:.3f}")
```

## Compare Multiple Models

```python
from tranpy.models import (
    DecisionTreeClassifier,
    RandomForestClassifier,
    LogisticRegression,
    GaussianNB
)

models = {
    'Decision Tree': DecisionTreeClassifier(max_depth=10),
    'Random Forest': RandomForestClassifier(n_estimators=50, max_depth=10),
    'Logistic Regression': LogisticRegression(max_iter=200),
    'Naive Bayes': GaussianNB()
}

results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    eval_results = model.evaluate(X_test, y_test, verbose=False)
    results[name] = eval_results['accuracy']
    print(f"{name:20s} Accuracy: {results[name]:.3f}")
```

## Save Model

```python
from tranpy.models import save_model

model = DecisionTreeClassifier(max_depth=10)
model.fit(X_train, y_train)

save_model(model, 'my_model.pkl')
```

## Load Model

```python
from tranpy.models import load_model

loaded_model = load_model('my_model.pkl')
predictions = loaded_model.predict(X_test[:5])

print(f"Predictions: {predictions}")
```
