# Using Pretrained Models

## List Available Models

```python
from tranpy.models import list_pretrained_models

models = list_pretrained_models()

print(f"Total models: {len(models)}\n")
for model in models[:5]:
    print(f"{model['model_id']:20s} {model['type']:30s} {model['grid']}")
```

## Filter by Grid

```python
ne_models = list_pretrained_models(grid='NewEngland')
nb_models = list_pretrained_models(grid='NineBusSystem')

print(f"NewEngland models: {len(ne_models)}")
print(f"NineBusSystem models: {len(nb_models)}")
```

## Model Types

```python
models = list_pretrained_models(grid='NewEngland')

for model in models:
    print(f"{model['model_id']:20s} {model['type']}")
```

## Load Pretrained Model

```python
from tranpy.models import load_pretrained
from tranpy.datasets import load_newengland

X_train, X_test, y_train, y_test = load_newengland(
    test_size=0.2,
    random_state=42
)

model = load_pretrained('dt_ne39', local=True)
predictions = model.predict(X_test[:10])

print(f"Predictions: {predictions}")
```

## Evaluate Pretrained Model

```python
model = load_pretrained('rf_ne39', local=True)
results = model.evaluate(X_test, y_test, verbose=False)

print(f"Accuracy: {results['accuracy']:.3f}")
print(f"\nConfusion Matrix:\n{results['confusion_matrix']}")
```

## Compare Pretrained Models

```python
model_ids = ['dt_ne39', 'rf_ne39', 'adaboost_ne39', 'lr_ne39']

results = {}
for model_id in model_ids:
    try:
        model = load_pretrained(model_id, local=True)
        eval_results = model.evaluate(X_test, y_test, verbose=False)
        results[model_id] = eval_results['accuracy']
        print(f"{model_id:20s} Accuracy: {results[model_id]:.3f}")
    except Exception as e:
        print(f"{model_id:20s} Error: {e}")
```

## Use with 9-Bus System

```python
from tranpy.datasets import load_ieee9bus

X_train, X_test, y_train, y_test = load_ieee9bus(
    test_size=0.2,
    random_state=42
)

model = load_pretrained('dt_9bus', local=True)
results = model.evaluate(X_test, y_test, verbose=False)

print(f"Accuracy: {results['accuracy']:.3f}")
```
