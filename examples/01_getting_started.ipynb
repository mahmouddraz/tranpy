# Getting started

Train and evaluate a stability classifier on the bundled New England 39-bus dataset.

```python
import tranpy
from tranpy_core import evaluate, make_pipeline, train_test_split
```

## Load the data

```python
data = tranpy.load_new_england()
data
```

## Split, fit, evaluate

The split is stratified and seeded; the pipeline scales features using the training rows only.

```python
split = train_test_split(data.labels, test_size=0.2, seed=42)
X_train, y_train = data.features[split.train], data.labels[split.train]
X_test, y_test = data.features[split.test], data.labels[split.test]

model = make_pipeline("random_forest", seed=0).fit(X_train, y_train)
scores = evaluate(model, X_test, y_test)
scores.accuracy, scores.f1, scores.roc_auc
```

```python
scores.confusion_matrix
```

## Compare a few models

```python
import pandas as pd

rows = {}
for name in ["dummy", "logistic_regression", "svm", "random_forest", "gradient_boosting"]:
    fitted = make_pipeline(name, seed=0).fit(X_train, y_train)
    rows[name] = evaluate(fitted, X_test, y_test).as_dict()
pd.DataFrame(rows).T[["accuracy", "balanced_accuracy", "f1", "roc_auc"]].round(4)
```
